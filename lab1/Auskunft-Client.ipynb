# Telefonauskunft – Client

Dieses Notebook ist die **Benutzerschnittstelle**. Der Server (`Auskunft-Server.ipynb`) muss bereits laufen.

Die Auskunft-API besteht aus zwei Methoden der Klasse `Client`:

- `get(name)` liefert die Telefonnummer zu einem Namen oder `None`.
- `getall()` liefert das ganze Telefonbuch als Dictionary.

## 1. Verbindung aufbauen

```python
import clientserver

client = clientserver.Client()  # connect() zum Server, siehe Log
```

## 2. Einzelnen Namen abfragen (`GET`)

```python
client.get("Bob")
```

Ein Name, der nicht im Telefonbuch steht, ergibt `None` (der Server antwortet mit `NOT_FOUND`):

```python
print(client.get("Niemand"))
```

## 3. Alle Einträge abfragen (`GETALL`)

```python
for name, number in client.getall().items():
    print(f"{name:<10} {number}")
```

## 4. Verbindung schließen

Erst danach kann der Server den nächsten Client bedienen (er bedient immer nur einen Client gleichzeitig).

```python
client.close()
```
