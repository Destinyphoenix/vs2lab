# Telefonauskunft – Server

Dieses Notebook startet den **Auskunft-Server**. Es muss **vor** dem Client-Notebook (`Auskunft-Client.ipynb`) laufen.

Der Server hält das Telefonbuch als Python-Dictionary im Speicher (Datenebene) und beantwortet über TCP-Sockets die Anfragen `GET` und `GETALL` (Auskunft-Protokoll, siehe Docstring in `clientserver.py`).

## 1. Telefonbuch festlegen

Ohne Argument verwendet der Server ein kleines Beispiel-Telefonbuch. Hier geben wir ein eigenes vor.

```python
import clientserver  # unser Modul mit Server, Client und Protokoll

phonebook = {
    "Alice": "0721 925-1234",
    "Bob": "0721 925-5678",
    "Charlie": "0721 925-9012",
    "Jürgen": "0721 925-3456",
    "test": "072177888",
}
```

## 2. Server erzeugen

Der Konstruktor erzeugt den Socket und bindet ihn an `127.0.0.1:50007`. Das sieht man in der Log-Ausgabe.

```python
server = clientserver.Server(phonebook)
```

## 3. Server starten

`serve()` blockiert: Die Zelle läuft, bis du den Kernel neu startest (das gibt auch den Port wieder frei). Jetzt das Client-Notebook ausführen. In den Logs hier siehst du dann die einzelnen Schritte: Verbindung annehmen, Anfrage empfangen, Antwort senden, Verbindung schließen.

```python
server.serve()
```
